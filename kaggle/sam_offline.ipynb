```python
import os
os.environ['OPERATION_MODE'] = 'offline'

# Install from Kaggle dataset (adjust path)
# !pip install -q /kaggle/input/arc-agi-3-wheels/*.whl torch pyyaml

import json
import sys
sys.path.insert(0, '/kaggle/working')

from sam.agent import SamAgent

agent = SamAgent(
    game_id='ls20-9607627b',
    environments_dir='/kaggle/input/arc-environments/environment_files',
    checkpoint='/kaggle/input/sam-checkpoints/global_sam.pt',
    run_dir='/kaggle/working/sam_runs',
    planning_ms=30.0,
)
result = agent.run()
print(json.dumps(result, indent=2))
```

```python
# RHAE dashboard
rhae = result.get('rhae', {})
print('Total RHAE:', rhae.get('total_score'))
for lv in rhae.get('level_scores', []):
    print(f"  L{lv['level']}: {lv['actions']} actions (baseline {lv['baseline']}) -> {lv['score']}%")
```
